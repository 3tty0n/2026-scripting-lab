## 課題1: 電圧を判定する (15点)

**やること**: 電圧 `v` (ボルト) を受け取り、範囲に応じて文字列を返す関数 `check_voltage(v)` を書く。

| 電圧 | 返す文字列 |
|---|---|
| 4.5 より小さい | `'低い'` |
| 4.5 以上 5.5 以下 | `'正常'` |
| 5.5 より大きい | `'高い'` |

**入出力の例**

| 呼び方 | 結果 |
|---|---|
| `check_voltage(3.3)` | `'低い'` |
| `check_voltage(5.0)` | `'正常'` |
| `check_voltage(4.5)` | `'正常'` (ちょうど 4.5 は正常) |
| `check_voltage(6.0)` | `'高い'` |

**手順**

1. `v < 4.5` なら `'低い'` を返す
2. そうでなくて `v <= 5.5` なら `'正常'` を返す
3. どれでもなければ `'高い'` を返す

**ヒント**: `if` → `elif` → `else` は上から順に調べ、最初に当てはまった所だけを実行する。

**つまずきやすい点**

- `<` と `<=` の取り違え → `check_voltage(4.5)` や `check_voltage(5.5)` が `'正常'` にならず NG になる
- `return` を書かずに `print` している → 結果が `None` になる

### 例題

次のセルを実行して、動きを見る。温度を 3 段階に分ける例。

```python
# 例題
def judge_temp(t):
    if t < 10:
        return '寒い'
    elif t <= 25:
        return 'ちょうどよい'
    else:
        return '暑い'


print(judge_temp(5), judge_temp(20), judge_temp(30))   # 寒い ちょうどよい 暑い
```

```python
def check_voltage(v):
    if v < 4.5:
        return '低い'
    elif v <= 5.5:
        return '正常'
    else:
        return '高い'


print(check_voltage(3.3), check_voltage(5.0), check_voltage(6.0))
```

```python
# [CHECK] q1
need('check_voltage', arity=1)

assert check_voltage(3.3) == '低い'
assert check_voltage(5.0) == '正常'
assert check_voltage(6.0) == '高い'
assert check_voltage(4.5) == '正常', 'ちょうど 4.5 は正常 (4.5 以上)'
assert check_voltage(5.5) == '正常', 'ちょうど 5.5 は正常 (5.5 以下)'
assert check_voltage(4.49) == '低い'
assert check_voltage(5.51) == '高い'
print('OK')
```

## 課題2: 合計と最大を求める (20点)

**やること**: 数のリスト `values` を受け取り、**合計**と**最大値**を組 (タプル) にして返す関数
`total_and_max(values)` を書く。`sum` や `max` は使わず、`for` で 1 つずつ見て求める。

- タプル: `(合計, 最大)` のように、値を 2 つまとめて返す形
- 空のリスト `[]` のときは `(0, None)` を返す (最大値が無いので `None`)

**入出力の例**

| 呼び方 | 結果 |
|---|---|
| `total_and_max([3, 1, 4])` | `(8, 4)` |
| `total_and_max([-5, -2])` | `(-7, -2)` |
| `total_and_max([])` | `(0, None)` |

**手順**

1. 空のリストなら `(0, None)` を返す (この部分は書いてある)
2. `total` を 0、`biggest` を先頭の値 `values[0]` から始める (書いてある)
3. `for` で 1 つずつ見て、`total` に足す
4. 今の `biggest` より大きければ、`biggest` を入れ替える

**ヒント**: スライドの「for で合計と最大を求める」とほぼ同じ形。

**つまずきやすい点**

- `biggest = 0` から始めている → 全部マイナスのリスト `[-5, -2]` で最大が 0 になってしまう
- `sum` / `max` を使っている → 「for で 1 つずつ足すこと」と出て NG になる
- `return` が `for` の中にある (字下げが深い) → 1 つ目を見ただけで終わってしまう

### 例題

次のセルを実行して、動きを見る。リストの中で 5 以上の値がいくつあるかを数える例。

```python
# 例題
def count_big(values):
    count = 0
    for v in values:
        if v >= 5:
            count = count + 1
    return count


print(count_big([3, 5, 8, 1]))   # 2
```

```python
def total_and_max(values):
    if len(values) == 0:
        return (0, None)
    total = 0
    biggest = values[0]
    for v in values:
        total = total + v
        if v > biggest:
            biggest = v
    return (total, biggest)


print(total_and_max([3, 1, 4]), total_and_max([]))
```

```python
# [CHECK] q2
need('total_and_max', arity=1)

assert total_and_max([3, 1, 4]) == (8, 4)
assert total_and_max([7]) == (7, 7)
assert total_and_max([-5, -2]) == (-7, -2), 'マイナスだけのリストでも最大が正しいこと (biggest は先頭から始める)'
assert total_and_max([]) == (0, None)
assert abs(total_and_max([3.3, 5.0, 4.8])[0] - 13.1) < 1e-9
holds('合計と最大が組み込み関数と同じ',
      lambda a, b, c: total_and_max([a, b, c]) == (a + b + c, max(a, b, c)), ints(3))
_t = tags('total_and_max')
assert not ({'calls:sum', 'calls:max'} & _t), 'sum / max を使わず、for で 1 つずつ足すこと'
print('OK')
```

## 課題3: 何回 2 倍すれば届くか (15点)

**やること**: 値 `start` を何回 2 倍すれば `limit` 以上になるかを返す関数 `count_doublings(start, limit)` を書く。
`while` (条件が成り立つ間くり返す) を使う。

- 最初から `start >= limit` なら、2 倍する必要がないので `0`
- `start` は 1 以上の数とする

**入出力の例**

| 呼び方 | 結果 | 理由 |
|---|---|---|
| `count_doublings(1, 100)` | `7` | 1→2→4→8→16→32→64→128 |
| `count_doublings(3, 10)` | `2` | 3→6→12 |
| `count_doublings(50, 10)` | `0` | 最初から 10 以上 |

**手順**

1. 回数 `count` を 0、今の値 `value` を `start` から始める (書いてある)
2. `while value < limit:` と書く (`value` が `limit` より小さい間くり返す)
3. その中 (字下げした行) で、`value` を 2 倍し、`count` を 1 増やす
4. 最後に `count` を返す (書いてある)

**ヒント**: スライドの「while: 条件が成り立つ間くり返す」と同じ形。

**つまずきやすい点**

- `while` の中で `value` を 2 倍し忘れる → いつまでも止まらない。確認セルが止まらないときは、メニューの `ランタイム → 実行を中断` を押す
- 条件を `value <= limit` にしている → ちょうど届いたとき (例: `count_doublings(1, 8)`) に 1 回多くなる

### 例題

次のセルを実行して、動きを見る。半分にしていって 1 以下になるまでの回数を数える例。

```python
# 例題
value = 100
count = 0
while value > 1:
    value = value / 2
    count = count + 1

print(count)   # 7
```

```python
def count_doublings(start, limit):
    count = 0
    value = start
    while value < limit:
        value = value * 2
        count = count + 1
    return count


print(count_doublings(1, 100), count_doublings(3, 10), count_doublings(50, 10))
```

```python
# [CHECK] q3
need('count_doublings', arity=2)

assert count_doublings(1, 100) == 7
assert count_doublings(3, 10) == 2
assert count_doublings(50, 10) == 0, '最初から limit 以上なら 0'
assert count_doublings(1, 8) == 3, 'ちょうど届いたら止める (条件は value < limit)'
assert count_doublings(1, 1) == 0
assert count_doublings(1, 1024) == 10
print('OK')
```

## 課題4: 辞書で数える (20点)

**やること**: 値のリスト `items` を受け取り、「値 → 出てきた回数」の辞書を返す関数 `count_items(items)` を書く。

**入出力の例**

| 呼び方 | 結果 |
|---|---|
| `count_items(['R', 'C', 'R'])` | `{'R': 2, 'C': 1}` |
| `count_items([])` | `{}` |
| `count_items([5, 5, 5])` | `{5: 3}` |

**手順**

1. 空の辞書 `counts = {}` から始める (書いてある)
2. `for` で 1 つずつ見る
3. もう辞書にあれば (`x in counts`)、回数を 1 増やす
4. まだ無ければ、回数 1 で追加する
5. 最後に `counts` を返す (書いてある)

**ヒント**: スライドの「辞書で数える」と同じ形。`counts[x] = counts[x] + 1` で 1 増やせる。

**つまずきやすい点**

- `in` で調べずに `counts[x] = counts[x] + 1` だけ書く → 初めての値で `KeyError` になる
- 渡されたリスト `items` そのものを書き換えている → 「引数を書き換えている」と出て NG になる

### 例題

次のセルを実行して、動きを見る。種類ごとに個数を足し合わせる例。

```python
# 例題
orders = [('R', 10), ('C', 5), ('R', 20)]

totals = {}
for kind, n in orders:
    if kind in totals:
        totals[kind] = totals[kind] + n
    else:
        totals[kind] = n

print(totals)   # {'R': 30, 'C': 5}
```

```python
def count_items(items):
    counts = {}
    for x in items:
        if x in counts:
            counts[x] = counts[x] + 1
        else:
            counts[x] = 1
    return counts


print(count_items(['R', 'C', 'R', 'L', 'R']))
```

```python
# [CHECK] q4
need('count_items', arity=1)

assert count_items(['R', 'C', 'R']) == {'R': 2, 'C': 1}
assert count_items([]) == {}
assert count_items([5, 5, 5]) == {5: 3}
assert count_items(list('abracadabra')) == {'a': 5, 'b': 2, 'r': 2, 'c': 1, 'd': 1}
holds('回数の合計は要素の数と同じ',
      lambda a, b, c: sum(count_items([a, b, c, a]).values()) == 4, ints(3, lo=0, hi=3))
unchanged('count_items', ['R', 'C', 'R'])
print('OK')
```

## 課題5: 逆ポーランド記法の電卓 (15点)

**やること**: `'3 4 +'` のような式の文字列を受け取り、計算結果を返す関数 `rpn(text)` を書く。

- **逆ポーランド記法**: 数を先に書き、記号 (`+` `-` `*`) を後に書く書き方。`3 4 +` は `3 + 4`
- **スタック**: 最後に積んだものから取り出す入れ物。リストの `append` (積む) と `pop` (取り出す) で作る
- 計算に使う関数は、辞書 `OPS` に入れてある (書いてある)

**入出力の例**

| 呼び方 | 結果 | スタックの動き |
|---|---|---|
| `rpn('3 4 +')` | `7` | `[3]` → `[3, 4]` → `[7]` |
| `rpn('10 3 -')` | `7` | `10 - 3` (先に取り出す 3 が右側) |
| `rpn('2 3 4 * +')` | `14` | `[2, 3, 4]` → `[2, 12]` → `[14]` |

**手順**

1. 空のスタック `stack = []` から始め、式を空白で区切って 1 つずつ見る (書いてある)
2. 記号なら (`t in OPS`)、上から 2 つ取り出す。先に取り出すのが `b` (右側)、次が `a` (左側)
3. `OPS[t](a, b)` で計算し、結果をスタックに積む
4. 記号でなければ数なので、`int(t)` で数に変えて積む
5. 最後に、スタックの一番上 `stack[-1]` を返す (書いてある)

**ヒント**: スライドの「逆ポーランド記法の電卓」と同じ形。

**つまずきやすい点**

- `a` と `b` を逆に取り出している → `rpn('10 3 -')` が `-7` になって NG になる
- 数を `int(t)` に変えずに積む → 文字列のまま計算され、`'3' + '4'` が `'34'` になる

### 例題

次のセルを実行して、動きを見る。スタックに積んで取り出す動き。

```python
# 例題
stack = []
stack.append(3)
stack.append(4)
print(stack)           # [3, 4]
top = stack.pop()
print(top, stack)      # 4 [3]
```

```python
def add(a, b):
    return a + b


def sub(a, b):
    return a - b


def mul(a, b):
    return a * b


OPS = {'+': add, '-': sub, '*': mul}


def rpn(text):
    stack = []
    for t in text.split():
        if t in OPS:
            b = stack.pop()
            a = stack.pop()
            stack.append(OPS[t](a, b))
        else:
            stack.append(int(t))
    return stack[-1]


print(rpn('3 4 +'), rpn('10 3 -'), rpn('2 3 4 * +'))
```

```python
# [CHECK] q5
need('rpn', arity=1)
need('OPS', kind=dict)

assert rpn('3 4 +') == 7
assert rpn('10 3 -') == 7, '先に取り出すのが右側の値 (10 3 - は 10 - 3)'
assert rpn('2 3 4 * +') == 14
assert rpn('5') == 5
assert rpn('1 2 + 3 *') == 9
holds('a b - は a - b', lambda a, b: rpn(f'{a} {b} -') == a - b, ints(2, lo=0, hi=50))
# 関数を辞書に足すだけで、新しい記号が使えること
OPS['max'] = max
assert rpn('3 9 max') == 9, '記号は OPS から引いて呼ぶこと'
del OPS['max']
print('OK')
```

## 発展1: 抵抗のカラーコード (10点)

**やること**: 抵抗の色の帯 3 本のリスト `colors` を受け取り、抵抗値 (オーム) を返す関数 `resistance(colors)` を書く。

- 色と数字の対応は辞書 `COLOR` に入れてある (書いてある)
- 1 本目と 2 本目で 2 桁の数を作り、3 本目の数だけ後ろに 0 を付ける
- 式で書くと `(10 × 1本目 + 2本目) × 10 の (3本目) 乗`
- 表に無い色があれば `ValueError` を出す

**入出力の例**

| 呼び方 | 結果 |
|---|---|
| `resistance(['茶', '黒', '赤'])` | `1000` (10 × 10²) |
| `resistance(['黄', '紫', '橙'])` | `47000` (47 × 10³) |
| `resistance(['茶', '金', '赤'])` | `ValueError` |

**手順**

1. 3 つの色それぞれについて、`COLOR` に無ければ `raise ValueError(...)` する
2. 3 つの色を `COLOR` で数字に変える
3. `(10 * d1 + d2) * 10 ** d3` を返す

**ヒント**: `10 ** 2` は 10 の 2 乗 (100)。

**つまずきやすい点**

- 調べずに `COLOR[c]` を引く → 表に無い色で `KeyError` になり、「ValueError を出すこと」と出る

```python
COLOR = {'黒': 0, '茶': 1, '赤': 2, '橙': 3, '黄': 4,
         '緑': 5, '青': 6, '紫': 7, '灰': 8, '白': 9}


def resistance(colors):
    for c in colors:
        if c not in COLOR:
            raise ValueError(f'知らない色: {c}')
    d1, d2, d3 = COLOR[colors[0]], COLOR[colors[1]], COLOR[colors[2]]
    return (10 * d1 + d2) * 10 ** d3


print(resistance(['茶', '黒', '赤']), resistance(['黄', '紫', '橙']))
```

```python
# [CHECK] b1
need('resistance', arity=1)

assert resistance(['茶', '黒', '赤']) == 1000
assert resistance(['黄', '紫', '橙']) == 47000
assert resistance(['赤', '赤', '黒']) == 22
try:
    resistance(['茶', '金', '赤'])
except ValueError:
    pass
else:
    raise AssertionError('表に無い色は ValueError を出すこと')
print('OK')
```

## 発展2: 両方にある部品 (5点)

**やること**: 部品の名前のリストを 2 つ受け取り、**両方に入っている名前**をアルファベット順に並べたリストで返す関数
`common_parts(a, b)` を書く。

**入出力の例**

| 呼び方 | 結果 |
|---|---|
| `common_parts(['R1', 'C1', 'R2'], ['R2', 'L1', 'R1'])` | `['R1', 'R2']` |
| `common_parts(['R1', 'R1'], ['R1'])` | `['R1']` (同じ名前は 1 つにまとめる) |
| `common_parts(['C1'], ['L1'])` | `[]` |

**手順**

1. 2 つのリストを集合にする (`set(a)`, `set(b)`)
2. `&` で両方にあるものを取り出す
3. `sorted(...)` で並べたリストにして返す

**つまずきやすい点**

- 集合のまま返している → 「並べたリストで返すこと」と出る

```python
def common_parts(a, b):
    return sorted(set(a) & set(b))


print(common_parts(['R1', 'C1', 'R2'], ['R2', 'L1', 'R1']))
```

```python
# [CHECK] b2
need('common_parts', arity=2)

r = common_parts(['R1', 'C1', 'R2'], ['R2', 'L1', 'R1'])
assert isinstance(r, list), '並べたリストで返すこと (sorted を使う)'
assert r == ['R1', 'R2']
assert common_parts(['R1', 'R1'], ['R1']) == ['R1'], '同じ名前は 1 つにまとめること'
assert common_parts(['C1'], ['L1']) == []
print('OK')
```
